```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Dataset manzili
data_path = 'sport_inshootlari.csv'

# Datasetni o'qish
df = pd.read_csv(data_path, encoding='utf-8')

# Datasetning statistik tavsifi
df.describe(include="all")
```

```python
# 1. Sport inshootlarini hududlar bo'yicha guruhlash
df_rank = df.groupby(["Hudud"])

# Har bir hudud bo'yicha sport inshootlari sonini chiqarish
print("Hududlar bo'yicha sport inshootlari soni:")
print(df_rank.size())
```

```python
# 2. "?" belgilarini NaN (bo'sh qiymat) bilan almashtirish
df.replace("?", np.nan, inplace=True)

# Har bir ustundagi bo'sh qiymatlarni tekshirish
print("\nHar bir ustundagi bo'sh qiymatlar:")
print(df.isnull().sum())
```

```python
# 3. T/r ustunini butun son (int) ma'lumot turiga o'tkazish
df["T/r"] = pd.to_numeric(df["T/r"], errors="coerce")
```

```python
# 4. Toshkent yoki boshqa muayyan hududdagi sport inshootlarini ajratish
df_sub = df[df["Hudud"] == "Jizzax viloyati"]

print("\nJizzax viloyatidagi sport inshootlari:")
display(df_sub)
```

```python
# Faqat kategorik ustunlarni tahlil qilish

df.describe(include=["object"])
```

```python
# Hududlar bo‘yicha sport inshootlari sonini hisoblash

df["Hudud"].value_counts()
```

```python
# Sport inshootlari turlari bo‘yicha soni

df["Sport inshoati turi"].value_counts()
```

```python
# Har bir tuman va sport inshooti turi bo'yicha sonini hisoblash
df_box = (
    df.groupby(["Tuman(shahar)", "Sport inshoati turi"])
      .size()
      .reset_index(name="Inshootlar soni")
)

# Boxplot grafik
plt.figure(figsize=(12, 6))

sns.boxplot(
    x="Sport inshoati turi",
    y="Inshootlar soni",
    data=df_box
)

# Grafik nomi va o'qlarini belgilash
plt.title("Sport inshootlari turlari bo'yicha sonlar taqsimlanishi")
plt.xlabel("Sport inshooti turi")
plt.ylabel("Inshootlar soni")

# X o'qi nomlarini qiya ko'rsatish
plt.xticks(rotation=45, ha="right")

plt.tight_layout()
plt.show()
```
