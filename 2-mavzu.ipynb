```python
from ucimlrepo import fetch_ucirepo
import pandas as pd

# Datasetni yuklash
automobile = fetch_ucirepo(id=10)

# DataFrame yaratish
df = automobile.data.features

# Tekshirish
print(df.head())
```

```python
# 1-qadam. Yonilg‘i turi kategorik o‘zgaruvchi ekanini tekshirish

# Yonilg'i turi ustunidagi noyob qiymatlar
print(df["fuel-type"].unique())

# Har bir yonilg'i turi nechta ekanini aniqlash
print(df["fuel-type"].value_counts())

# Ustunning ma'lumot turini tekshirish
print(df["fuel-type"].dtype)
```

```python
# 2-qadam. O‘zgaruvchini raqamli formatga o‘tkazishga tayyorlash

df_encoded = df.copy()
```

```python
# 3-qadam. Ikkita yangi xususiyat yaratish

import pandas as pd

df_encoded = pd.get_dummies(
    df_encoded,
    columns=["fuel-type"],
    dtype=int
)

print(df_encoded.columns.tolist())
```

```python
# 4-qadam. Mos keladigan qiymatni 1 ga, qolganini 0 ga tenglashtirish

print(
    df_encoded[
        ["fuel-type_diesel", "fuel-type_gas"]
    ].head(10)
)
```

```python
# 5-qadam. B avtomobilining dizelda ishlashini tekshirish
print(df.iloc[1]["fuel-type"])
```

```python

# Datasetdan dizelda ishlaydigan birinchi avtomobilni topish
diesel_car = df[df["fuel-type"] == "diesel"].iloc[[0]]

# Uning kodlangan qiymatlarini ko'rsatish
print(
    df_encoded.loc[
        diesel_car.index,
        ["fuel-type_diesel", "fuel-type_gas"]
    ]
)
```

```python
# 6-qadam. Kodlash natijasini tekshirish

print(df_encoded[
    ["fuel-type_diesel", "fuel-type_gas"]
].sum())
```

```python
# 7-qadam. Tayyor DataFrame'ni saqlash

df_encoded.to_csv(
    "automobile_encoded.csv",
    index=False
)
```
