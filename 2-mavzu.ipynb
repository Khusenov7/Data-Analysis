```python
import pandas as pd

url = "https://gist.githubusercontent.com/sachinsdate/2c31461b948e2695b2553a2c398aff3c/raw/automobiles.data.csv"

df = pd.read_csv(url)

print(df.head())
print(df.shape)
```

```python
# 1. Yonilg'i turi kategorik o'zgaruvchi ekanini tekshirish

# Yonilg'i turi ustunidagi noyob qiymatlar
print(df["fuel_type"].unique())

# Har bir yonilg'i turi nechta ekanini aniqlash
print(df["fuel_type"].value_counts())

# Ustunning ma'lumot turini tekshirish
print(df["fuel_type"].dtype)
```

```python
# 2. O'zgaruvchini raqamli formatga o'tkazishga tayyorlash

df_encoded = df.copy()
```

```python
# 3. Yonilg'i turini ikkita yangi xususiyatga ajratish

import pandas as pd

df_encoded = pd.get_dummies(
    df.copy(),
    columns=["fuel_type"],
    dtype=int
)

print(df_encoded.columns.tolist())
print(df_encoded.head())
```

```python
# 4. Mos keladigan qiymatni 1 ga, qolganini 0 ga tenglashtirish

print(
    df_encoded[
        ["fuel_type_diesel", "fuel_type_gas"]
    ].head(10)
)
```

```python
# 5. B avtomobilining dizelda ishlashini tekshirish
print(df.iloc[1]["fuel_type"])
```

```python

# Datasetdan dizelda ishlaydigan birinchi avtomobilni topish
diesel_car = df[df["fuel_type"] == "diesel"].iloc[[0]]

# Uning kodlangan qiymatlarini ko'rsatish
print(
    df_encoded.loc[
        diesel_car.index,
        ["fuel_type_diesel", "fuel_type_gas"]
    ]
)
```

```python
# 6. Kodlash natijasini tekshirish

print(df_encoded[
    ["fuel_type_diesel", "fuel_type_gas"]
].sum())
```

```python
# 7. Tayyor DataFrame'ni saqlash

df_encoded.to_csv(
    "automobile_encoded.csv",
    index=False
)
```
